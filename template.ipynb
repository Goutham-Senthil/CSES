### Test Case functions below. Do not edit.

```python
### Libraries
import time
import tracemalloc
import sys
import io
import glob
```

```python
def run_io_fun(inp, func, time_limit_sec=None, memory_limit_mb=None):
    old_stdin, old_stdout = sys.stdin, sys.stdout
    sys.stdin = io.StringIO(inp)
    sys.stdout = io.StringIO()
    
    start_time = time.time()
    tracemalloc.start()
    status = "Completed"
    peak = 0

    try:
        func()
        output = sys.stdout.getvalue()
        _, peak = tracemalloc.get_traced_memory()
        if memory_limit_mb is not None and (peak / (1024*1024)) > memory_limit_mb:
            status = "Memory limit exceeded"
    except Exception as e:
        output = ""
        status = f"Error: {e}"
    finally:
        elapsed_time = time.time() - start_time
        tracemalloc.stop()
        sys.stdin, sys.stdout = old_stdin, old_stdout

    if time_limit_sec is not None and elapsed_time > time_limit_sec:
        status = "Time limit exceeded"

    return output, elapsed_time, peak / (1024*1024), status
```

```python
def test_custom(inp: str, time_limit_sec=None, memory_limit_mb=None):
    out, t, mem, status = run_io_fun(inp, solve, time_limit_sec, memory_limit_mb)
    print("Output:")
    print(out)
    print(f"Time: {t:.6f} s, Memory: {mem:.3f} MB, Status: {status}")
```

```python
def test_custom_file(test_file, time_limit_sec=None, memory_limit_mb=None):
    """Test from a file and compare with expected output"""
    with open(test_file) as fin:
        inp = fin.read()
    
    got, t, mem, status = run_io_fun(inp, solve, time_limit_sec, memory_limit_mb)
    got = got.strip()
    
    with open(test_file.replace(".in", ".out")) as fout:
        exp = fout.read().strip()
    
    if got == exp and status == "Completed":
        print(f"{test_file} ✅ Time: {t:.6f} s, Memory peak: {mem:.3f} MB")
    else:
        print(f"{test_file} ❌ Status: {status}")
        print(f"Expected:\n{exp}")
        print(f"Got:\n{got}")
        print(f"Time: {t:.6f} s, Memory peak: {mem:.3f} MB")
```

```python
def test_all(time_limits=None, memory_limits=None):
    """
    Runs all test files in tests/*.in
    time_limits and memory_limits are dictionaries mapping filename to limit.
    """
    for f in sorted(glob.glob("tests/*.in")):
        with open(f) as fin:
            inp = fin.read()

        # Set per-file limits if provided, else None
        t_limit = time_limits.get(f, None) if time_limits else None
        m_limit = memory_limits.get(f, None) if memory_limits else None

        got, t, mem, status = run_io_fun(inp, solve, t_limit, m_limit)
        got = got.strip()

        with open(f.replace(".in", ".out")) as fout:
            exp = fout.read().strip()

        if got == exp and status == "Completed":
            print(f"{f} ✅ Time: {t:.6f} s, Memory peak: {mem:.3f} MB")
        else:
            print(f"{f} ❌ Status: {status}")
            print(f"Expected:\n{exp}")
            print(f"Got:\n{got}")
            print(f"Time: {t:.6f} s, Memory peak: {mem:.3f} MB")
            return

    print("All tests passed 🎉")
```
