```python
from dataclasses import dataclass
from typing import List


@dataclass
class Sample:
    t: float
    stack: List[str]


def reconstruct_trace(samples: List[Sample]) -> List[str]:
    """
    Given a list of sampling profiler snapshots ordered by timestamp,
    reconstruct the most likely sequence of function call and return events.

    Each Sample contains:
      - t: timestamp
      - stack: call stack at that moment

    The stack is ordered from outermost function to innermost currently
    executing function.

    An arbitrary number of function calls and returns may occur between
    consecutive samples.

    Return a list of strings representing events in chronological order.

    Example event strings:
        "call foo"
        "return foo"
    """
    events = []

    prev_stack = []


    for t,curr_stack in [(s.t, s.stack) for s in samples ]:
        i = 0 


        # this is basically 
        # finding the first point of 
        # difference 
        # so anything 
        # at the difference point 
        # so everyting after that ended and everything after that started
        while (
            i<len(curr_stack)
            and i<len(prev_stack)
            and curr_stack[i] == prev_stack[i]
        ):
            i+=1

        # functions that have returned 
        for func in prev_stack[i:]:
            events.append(f"return {func} at time {t}")
        
        for func in curr_stack[i:]:
            events.append(f"call {func} at time {t}")
        
        prev_stack = curr_stack
    
    return events
                             
        


samples = [
    Sample(0.0, ["main"]),
    Sample(1.0, ["main", "foo"]),
    Sample(2.0, ["main", "foo", "bar","main"]),
    Sample(3.0, ["main", "baz"]),
]
```

```python
def binary_search(nums, target):
    l, r = 0, len(nums) - 1

    while l <= r:
        mid = l + (r - l) // 2

        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            l = mid + 1
        else:
            r = mid - 1

    return -1
```

```python
import tkinter as tk
import random

class DragSortApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Drag Sort Game")

        self.canvas_width = 900
        self.canvas_height = 400

        self.main_frame = tk.Frame(root)
        self.main_frame.pack()

        self.show_menu()

    # ---------------- MENU ----------------
    def show_menu(self):
        self.clear_frame()

        tk.Label(self.main_frame, text="Drag Sort Game", font=("Arial", 24, "bold")).pack(pady=20)

        tk.Button(self.main_frame, text="Easy (5)", width=25,
                  command=lambda: self.start_game(5)).pack(pady=5)
        tk.Button(self.main_frame, text="Medium (15)", width=25,
                  command=lambda: self.start_game(15)).pack(pady=5)
        tk.Button(self.main_frame, text="Hard (30)", width=25,
                  command=lambda: self.start_game(30)).pack(pady=5)
        tk.Button(self.main_frame, text="Stupidly Hard (100,000)", width=25,
                  command=lambda: self.start_game(100_000)).pack(pady=5)

    # ---------------- GAME ----------------
    def start_game(self, size):
        self.clear_frame()

        self.size = size
        self.showing_solution = False

        if size < 1000:
            # small array → draw all
            self.data = random.sample(range(1, size * 3), size)
            self.virtualized = False
        else:
            # huge array → virtualized
            self.data = random.sample(range(1, size * 2), size)
            self.virtualized = True
            self.view_start = 0
            self.view_size = 150  # only render 150 bars

        # Canvas
        self.canvas = tk.Canvas(self.main_frame,
                                width=self.canvas_width,
                                height=self.canvas_height,
                                bg="white")
        self.canvas.pack()

        # Buttons
        btn_frame = tk.Frame(self.main_frame)
        btn_frame.pack(pady=5)

        self.solution_btn = tk.Button(btn_frame, text="Show Sorted", command=self.toggle_solution)
        self.solution_btn.grid(row=0, column=0, padx=5)

        tk.Button(btn_frame, text="Back to Menu", command=self.show_menu).grid(row=0, column=1, padx=5)

        # Scrollbar for huge arrays
        if self.virtualized:
            self.scroll = tk.Scale(self.main_frame, from_=0, to=self.size - self.view_size,
                                   orient="horizontal", length=800, command=self.on_scroll)
            self.scroll.pack()

        self.items = []
        self.dragging_index = None
        self.drag_item = None
        self.offset_x = 0

        self.draw()

    # ---------------- DRAW ----------------
    def draw(self):
        self.canvas.delete("all")
        self.items.clear()

        if self.virtualized:
            visible = self.data[self.view_start:self.view_start + self.view_size]
            n = len(visible)
            width = self.canvas_width // n
            max_val = max(visible)

            for i, value in enumerate(visible):
                x0 = i * width
                x1 = (i + 1) * width
                h = (value / max_val) * self.canvas_height
                y0 = self.canvas_height - h

                rect = self.canvas.create_rectangle(x0, y0, x1, self.canvas_height, fill="#4CAF50", outline="")
                self.canvas.tag_bind(rect, "<ButtonPress-1>", self.start_drag)
                self.canvas.tag_bind(rect, "<B1-Motion>", self.drag)
                self.canvas.tag_bind(rect, "<ButtonRelease-1>", self.drop)
                self.items.append(rect)
        else:
            # draw everything for small arrays
            n = len(self.data)
            width = self.canvas_width // n
            max_val = max(self.data)

            for i, value in enumerate(self.data):
                x0 = i * width + 2
                x1 = (i + 1) * width - 2
                h = (value / max_val) * self.canvas_height
                y0 = self.canvas_height - h

                rect = self.canvas.create_rectangle(x0, y0, x1, self.canvas_height, fill="#4CAF50", outline="")
                text = None
                if n <= 15:
                    text = self.canvas.create_text((x0 + x1)//2, y0 - 10, text=str(value), font=("Arial", 10, "bold"))
                self.canvas.tag_bind(rect, "<ButtonPress-1>", self.start_drag)
                self.canvas.tag_bind(rect, "<B1-Motion>", self.drag)
                self.canvas.tag_bind(rect, "<ButtonRelease-1>", self.drop)
                self.items.append((rect, text))

        # Draw solution overlay if requested
        if self.showing_solution:
            sorted_data = sorted(self.data)
            if self.virtualized:
                for i, value in enumerate(self.data[self.view_start:self.view_start + self.view_size]):
                    x0 = i * width
                    x1 = (i + 1) * width
                    h = (value / max(visible)) * self.canvas_height
                    y0 = self.canvas_height - h
                    self.canvas.create_rectangle(x0, y0, x1, self.canvas_height,
                                                 fill="#2196F3", stipple="gray50", outline="")
            else:
                for i, value in enumerate(sorted_data):
                    x0 = i * width + 6
                    x1 = (i + 1) * width - 6
                    h = (value / max(self.data)) * self.canvas_height
                    y0 = self.canvas_height - h
                    self.canvas.create_rectangle(x0, y0, x1, self.canvas_height,
                                                 fill="#2196F3", stipple="gray50", outline="")

    # ---------------- SCROLL ----------------
    def on_scroll(self, val):
        self.view_start = int(val)
        self.draw()

    # ---------------- DRAG ----------------
    def start_drag(self, event):
        if self.showing_solution:
            return

        item = self.canvas.find_closest(event.x, event.y)[0]

        if self.virtualized:
            local_index = self.items.index(item)
            self.dragging_index = self.view_start + local_index
        else:
            self.dragging_index = self.items.index(next(x for x in self.items if x[0]==item))

        self.drag_item = item
        self.offset_x = event.x
        self.canvas.itemconfig(item, fill="orange")

    def drag(self, event):
        if not self.drag_item:
            return

        dx = event.x - self.offset_x
        self.canvas.move(self.drag_item, dx, 0)
        self.offset_x = event.x

    def drop(self, event):
        if self.dragging_index is None:
            return

        if self.virtualized:
            width = self.canvas_width // self.view_size
            new_local_index = max(0, min(self.view_size - 1, event.x // width))
            new_index = self.view_start + new_local_index
        else:
            n = len(self.data)
            width = self.canvas_width // n
            new_index = max(0, min(n - 1, event.x // width))

        value = self.data.pop(self.dragging_index)
        self.data.insert(new_index, value)

        self.drag_item = None
        self.dragging_index = None
        self.draw()

        if self.data[:self.view_size] == sorted(self.data[:self.view_size]):
            print("Sorted portion looks correct! 🎉")

    # ---------------- UTIL ----------------
    def toggle_solution(self):
        self.showing_solution = not self.showing_solution
        self.draw()

    def clear_frame(self):
        for widget in self.main_frame.winfo_children():
            widget.destroy()


if __name__ == "__main__":
    root = tk.Tk()
    app = DragSortApp(root)
    root.mainloop()
```

```python
from typing import List, Tuple

def kadanes(nums: List[int]) -> Tuple[List[int], int]:
    if not nums:
        return [], 0

    curSum = nums[0]
    maxSum = nums[0]
    
    l = 0
    L = 0
    R = 0

    for r in range(1, len(nums)):
        if curSum < 0:
            curSum = nums[r]
            l = r
        else:
            curSum += nums[r]

        if curSum > maxSum:
            maxSum = curSum
            L = l
            R = r

    return nums[L:R+1], maxSum
```

```python
def maxSubarraySum(arr):
    res = arr[0]
  
    # Outer loop for starting point of subarray
    for i in range(len(arr)):
        currSum = 0
      
        # Inner loop for ending point of subarray
        for j in range(i, len(arr)):
            currSum = currSum + arr[j]
          
            # Update res if currSum is greater than res
            res = max(res, currSum)
          
    return res
```

```python
def hanoi(n,start,end):
    """
        n -> Number of disks
        start -> (1,2,3) referring to what starting stick 
        end -> (1,2,3) referring to what ending stick

        Some assumptions
        start =/= end
        n >= 1
    """
    if n == 1:
        print(f"{start} => {end}")
        return

    other = 6 - (start + end)
    hanoi(n - 1, start , other)
    print(f"{start} => {end}")
    hanoi(n - 1, other , end)

    return
```
